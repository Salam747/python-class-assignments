```python
class Library():

  #constructor
  def __init__(self, name, location):

    #class Attributes/ Class Varibales/ Parameters
    self.books = []
    self.lib_name = name
    self.lib_loacation = location

  #Class methods / Function
  def add_book(self, book):
    return self.books.append(book)

  def remove_book(self, remove_book):
    return self.books.remove(remove_book)

  def get_book(self):
    return self.books

# Making an object / Creating instance of a class
my_lib = Library("The Human ", "Karachi Pakistan")
print(my_lib.lib_name)
print(my_lib.lib_loacation)




class Book():
  def __init__(self, name, author, availability):
    self.book_name = name
    self.author_name = author
    self.avail = availability

  def borrow(self):
    self.avail





# Instance of class moved outside the class definition
#book1 = Book("Poetry", "Allama Iqbal", True) # Moved outside the Book class
#print(book1.book_name)
#print(book1.author_name)
#print(book1.avail)


book2 = Book("Khan", "Imran Khan", True)
book3 = Book("Naya pakistan", "Imran Khan", True)

mylib2 = Library("The Libraray", "Karachi")

# adiing books to the list in Libraray
print(mylib2.add_book(book2.book_name))
print(mylib2.add_book(book3.book_name))


# Getting the list of books In Libraray
print(mylib2.get_book())


#Removing the book in  Library
print(mylib2.remove_book("Khan"))

#Fetching the list again after removing
print(mylib2.get_book())
```

```python
class Hospital:
    def __init__(self, name, location):
        self.name = name
        self.location = location
        self.doctors = []
        self.patients = []

    def add_doctor(self, doctor):
        self.doctors.append(doctor)

    def remove_doctor(self, doctor_name):
        for doctor in self.doctors:
            if doctor.name == doctor_name:
                self.doctors.remove(doctor)
                return
        print(f"Doctor '{doctor_name}' not found in {self.name}")

    def add_patient(self, patient):
        self.patients.append(patient)

    def remove_patient(self, patient_name):
        for patient in self.patients:
            if patient.name == patient_name:
                self.patients.remove(patient)
                return
        print(f"Patient '{patient_name}' not found in {self.name}")

    def display_doctors(self):
        print("Hospital Doctors:")
        for doctor in self.doctors:
            print(f"- Name: {doctor.name}, Specialization: {doctor.specialization}")

    def display_patients(self):
        print("Hospital Patients:")
        for patient in self.patients:
            print(f"- Name: {patient.name}, Condition: {patient.condition}")



class Doctor:
    def __init__(self, name, specialization):
        self.name = name
        self.specialization = specialization

class Patient:
    def __init__(self, name, condition):
        self.name = name
        self.condition = condition


# Create a hospital instance
hospital = Hospital("City Hospital Karachi", "Koeangi 4 Main St")

# Create doctor and patient instances
doctor1 = Doctor("Dr. Abdul Salam", "Diagnostician")
doctor2 = Doctor("Dr. Sameer", "Oncology")
patient1 = Patient("Zahid", "Broken Leg")
patient2 = Patient("Saif", "Flu")


# Add doctors and patients to the hospital
hospital.add_doctor(doctor1)
hospital.add_doctor(doctor2)
hospital.add_patient(patient1)
hospital.add_patient(patient2)


# Display hospital information using methods
hospital.display_doctors()
hospital.display_patients()
```
