```python
# POSITIONAL ARGUMENTS
# The order of arguments passed must match the order of parameters defined in the function

def book_ticket(name, destination, date):
    print(f"Booking a ticket for {name} to {destination} on {date}.")

book_ticket("Abdul Salam", "China", "6th April 2025")
```

```python
# KEYWORD ARGUMENTS

# in this case, argumnets are passed using the paramters names,
# and we can specify them in any order

def book_ticket(name, destination, date):
    print(f"Booking a ticket for {name} to {destination} on {date}")

#function call with keyword argumnets
book_ticket(destination="Turkey", name="Abdul Salam", date="9th April 2025")
```

```python
def check(cart_id, *, customer_name, total_amount, payment_method):
    order ={
        "cart_id": cart_id,
        "customer": customer_name,
        "total_amount": total_amount,
        "paymnet": payment_method
    }
    print(order)

check("CART433", customer_name= "Abdul Salam", total_amount= 4700, payment_method="Credit Card")
check("CART433", payment_method="Credit Card",customer_name="Abdul Salam", total_amount=4700)
```
